```python
from pathlib import Path
import shutil

TEST_IMGS = Path('/kaggle/input/datasets/rondii/yolo-rtsd/rtsd_yolo/images/test')
TEST_LBLS = Path('/kaggle/input/datasets/rondii/yolo-rtsd/rtsd_yolo/labels/test')
names = ['3_24', '2_4', '3_27', '1_23', '4_1_1', '5_20']

false_negatives = []
false_positives = []
wrong_class     = []
low_conf        = []

for img_path in TEST_IMGS.glob('*.jpg'):
    lbl_path = TEST_LBLS / (img_path.stem + '.txt')
    if not lbl_path.exists():
        continue

    gt_classes = set()
    for line in lbl_path.read_text().splitlines():
        if line.strip():
            gt_classes.add(int(line.split()[0]))

    r = model(str(img_path), verbose=False, conf=0.1)[0]
    pred_classes = {}
    for box in r.boxes:
        cls  = int(box.cls)
        conf = float(box.conf)
        pred_classes[cls] = conf

    for cls in gt_classes:
        if cls not in pred_classes:
            false_negatives.append((img_path, names[cls]))

    for cls, conf in pred_classes.items():
        if cls not in gt_classes and conf > 0.3:
            false_positives.append((img_path, names[cls], conf))

    for cls, conf in pred_classes.items():
        if cls in gt_classes and conf < 0.5:
            low_conf.append((img_path, names[cls], conf))

print(f'Пропустила: {len(false_negatives)}')
print(f'Галлюцинации: {len(false_positives)}')
print(f'Низкая уверенность: {len(low_conf)}')
print(f'Неправильный класс: {len(wrong_class)}')
```

```python
hard_imgs = list(dict.fromkeys(str(p) for p, _, _ in low_conf))
fail_imgs = list(dict.fromkeys(
    [str(p) for p, _ in false_negatives[:2]] +
    [str(p) for p, _, _ in sorted(false_positives, key=lambda x: x[2])[:3]]
))

model(hard_imgs, conf=0.1, save=True,
      project='/kaggle/working', name='hard_cases')

model(fail_imgs, conf=0.1, save=True,
      project='/kaggle/working', name='failures')

print('Hard cases:', len(hard_imgs))
print('Failures:', len(fail_imgs))
```
